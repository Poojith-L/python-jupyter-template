### Exercise 6: Histogram Customization

TODO: Create a histogram for the following data: data = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4, 5].
Customize the histogram with a title, labels for the x-axis, and a specific color for the bars.

```python
import matplotlib.pyplot as plt
import numpy as np

# Data for the histogram
data = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4, 5]

# Create histogram data
counts, bins = np.histogram(data, bins=5)

# Define the colors for each bar
colors = ['red', 'blue', 'green', 'purple', 'orange']

# Create the bars using plt.bar() to specify color for each bin
plt.bar(bins[:-1], counts, width=np.diff(bins), align='edge', color=colors)

# Add title and labels
plt.title('Colourfull Histogram')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Show the plot
plt.show()
```
